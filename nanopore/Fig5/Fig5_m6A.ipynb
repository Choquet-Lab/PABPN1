```python
def position_m6A_UTR(strand, site, start, stop) :
    if strand ==  '+' :
        if site > stop :
            return('3UTR')
        elif (site < stop) & (site > start) :
            return('CDS')
        elif site < start :
            return('5UTR')
        else :
            return('unknown')
    elif strand == '-' :
        if site < stop :
            return('3UTR')
        elif (site > stop) & (site < start) :
            return('CDS')
        elif site > start :
            return('5UTR')
        else :
            return('unknown')

cyto_diff_m6A_merge_reps_cov_genes['position'] = cyto_diff_m6A_merge_reps_cov_genes.apply(lambda row : position_m6A_UTR(row.strand, row.start, row.start_codon, row.stop_codon), axis = 1)
```

```python
# get normalized position for the heatmap and metagene plot
def normalized_position(site, start_gene, start_codon, end_gene, stop_codon, position, strand) :
    if strand == '+' :
        if position == '5UTR' :
            return((site - start_gene) / (start_codon - start_gene))
        elif position == 'CDS' :
            return(1 + (site - start_codon) / (stop_codon - start_codon))
        elif position == '3UTR' :
            return(2 + (site - stop_codon) / (end_gene - stop_codon))
    elif strand == '-' :
        if position == '5UTR' :
            return((end_gene - site) / (end_gene - start_codon))
        elif position == 'CDS' :
            return(1 + (start_codon - site) / (start_codon - stop_codon))
        elif position == '3UTR' :
            return(2 + (stop_codon - site) / (stop_codon - start_gene))

cyto_diff_m6A_merge_reps_cov_genes['normalized_position'] = cyto_diff_m6A_merge_reps_cov_genes.apply(lambda row : normalized_position(row.start, row.start_gene, row.start_codon, row.end_gene, row.stop_codon, row.position, row.strand), axis = 1)
cyto_diff_m6A_merge_reps_cov_genes_more_both = cyto_diff_m6A_merge_reps_cov_genes[(cyto_diff_m6A_merge_reps_cov_genes['dir_x'] == 'more_modified') & (cyto_diff_m6A_merge_reps_cov_genes['dir_y'] == 'more_modified')]

cyto_diff_m6A_merge_reps_cov_genes['m6A_diff'] = 'no_change'
cyto_diff_m6A_merge_reps_cov_genes.loc[(cyto_diff_m6A_merge_reps_cov_genes['dir_x'] == 'more_modified') & (cyto_diff_m6A_merge_reps_cov_genes['dir_y'] == 'more_modified'), 'm6A_diff'] = 'more_methylated'

# keep sites modified at least at 5 % in one condition
cyto_diff_m6A_merge_reps_cov_genes_nonull = cyto_diff_m6A_merge_reps_cov_genes[(cyto_diff_m6A_merge_reps_cov_genes['a_pct_modified_x'] >= 0.05) | (cyto_diff_m6A_merge_reps_cov_genes['b_pct_modified_x'] >= 0.05) |
                                                                               (cyto_diff_m6A_merge_reps_cov_genes['a_pct_modified_y'] >= 0.05) | (cyto_diff_m6A_merge_reps_cov_genes['b_pct_modified_y'] >= 0.05)]

cyto_diff_m6A_merge_reps_cov_genes_nonull.head()
```

```python
# function to get reads-genes-m6A-polya association

def reads_polya_m6A_gene(reads_polya_gene_file, reads_m6A_file) :
    reads_polya_df = pd.read_table(reads_polya_gene_file, dtype={0:'str', 7:'str'}, header = None)
    reads_polya_df.columns = ['chrom', 'start', 'end', 'read_id', 'score', 'strand', 'polya_length', 'chrom_gene', 'start_gene', 'end_gene', 'Geneid', 'score_gene', 'strand_gene', 'length']
    reads_m6A_df = pd.read_table(reads_m6A_file, dtype={3:'str'}, header = None)
    reads_m6A_df.columns = ['read_id', 'forward_read_position', 'ref_position', 'chrom', 'mod_strand', 'ref_strand', 'ref_mod_strand', 'fw_soft_clipped_start', 'fw_soft_clipped_end', 'read_length',
          'call_prob', 'call_code', 'base_qual', 'ref_kmer', 'query_kmer', 'canonical_base', 'modified_primary_base', 'fail', 'inferred', 'within_alignment', 'flag']
    reads_polya_m6A_df = reads_polya_df.merge(reads_m6A_df, on=['chrom', 'read_id'], how='left')
    reads_polya_m6A_df_count = reads_polya_m6A_df.groupby(['read_id', 'Geneid', 'strand', 'polya_length'])['call_code'].count().reset_index()
    reads_polya_m6A_df_count['containing_m6A'] = 'No'
    reads_polya_m6A_df_count.loc[reads_polya_m6A_df_count['call_code'] > 0, 'containing_m6A'] = 'Yes'
    if 'cyto' in reads_polya_gene_file :
        reads_polya_m6A_df_count['compartment'] = 'cytoplasm'
    elif 'nucleus' in reads_polya_gene_file :
        reads_polya_m6A_df_count['compartment'] = 'nucleus'
    if 'rep4' in reads_polya_gene_file :
        reads_polya_m6A_df_count['replicate'] = 'rep4'
    else :
        reads_polya_m6A_df_count['replicate'] = 'rep3'
    if 'shPABPN1' in reads_polya_gene_file :
        reads_polya_m6A_df_count['shRNA'] = 'shPABPN1'
    elif 'shCTL' in reads_polya_gene_file :
        reads_polya_m6A_df_count['shRNA'] = 'shCTL'
    return(reads_polya_m6A_df, reads_polya_m6A_df_count)
```
